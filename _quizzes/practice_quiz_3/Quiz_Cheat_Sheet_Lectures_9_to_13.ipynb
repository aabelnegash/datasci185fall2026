## 2. QUICK CARD — Wording, operators, indentation, and data shapes

| Phrase | Comparison |
|---|---|
| at least 40 | `value >= 40` |
| more than / strictly greater than 40 | `value > 40` |
| at most / no more than 2 | `value <= 2` |
| below / less than 2 | `value < 2` |
| between 0 and 10, inclusive | `0 <= value <= 10` |
| equal / not equal | `==` / `!=` |

- `=` assigns a value or supplies a default/keyword argument; `==` asks whether two values are equal.
- `**` is exponentiation: write `x**2`, **not** `x^2`.
- `and` combines Boolean conditions; both must be true. `or` needs at least one true condition.
- In `for name in records:`, `name` is a **dictionary key**. `records[name]` is its value. `records[name]["score"]` goes one level deeper.
- A whole dictionary is not a single quantity. Convert `int(order[item])`, not `int(order)`.
- A two-point problem has **two separate tuples**. Each tuple contains both components of one point.
- `total += cost` accumulates; `total = cost` overwrites.
- Reset once per group: **inside the outer loop, before the inner loop**. A single overall total starts before the outer loop.
- `return` ends the function immediately. A final accumulated result normally belongs **after the loop**, indented inside the function.
- A `def` cell defines a function; call it to execute its body. In a notebook rerun a changed definition before testing its calls.
- A bare string inside a function does not display a message; use `print(...)`.
- `round(value, 2)` rounds a number. `f"{value:.2f}"` displays **exactly two digits** after the decimal, including zeros.

[Back to contents](#contents)


```python
# Small data-shape example: keys, records, and individual values.
records = {"Alex": {"score": 8.5, "year": 2004},
           "Jo": {"score": 7.0, "year": 2002}}
for name in records:
    print("key:", name, "record:", records[name], "score:", records[name]["score"])

amount = 80.0
print("Rounded number:", round(amount, 2))
print(f"Exactly two decimals: {amount:.2f}")
```

## 3. QUICK COPY — Running extrema and a matching name

**Use when:** find the oldest / earliest / weakest / largest / best entry and report who it belongs to.

Choose `<` for a minimum, `>` for a maximum. Update the number and its matching key **in the same `if` block**. Starting a minimum at `0` fails when every candidate is positive; starting a maximum at `0` fails for all-negative values. `float("inf")` / `float("-inf")` handle finite numeric values without assuming their sign. Empty data needs an explicit policy (the example below leaves the name as `None`). With strict comparisons, a tie keeps the first record encountered; `<=` or `>=` would replace it with the last tied record.

[Back to contents](#contents)


```python
roster = {"Alex": {"year": 2004, "skill": 8.5},
          "Jo": {"year": 2002, "skill": 7.0},
          "Sam": {"year": 2005, "skill": 9.0}}

# MINIMUM year = oldest person.
oldest_year = float("inf")
oldest_name = None
for name in roster:
    year = roster[name]["year"]
    if year < oldest_year:
        oldest_year = year
        oldest_name = name
print("Oldest:", oldest_name, oldest_year)

# MAXIMUM skill = strongest person.
best_skill = float("-inf")
best_name = None
for name in roster:
    skill = roster[name]["skill"]
    if skill > best_skill:
        best_skill = skill
        best_name = name
print("Best:", best_name, best_skill)
```

## 4. QUICK COPY — Nested-loop grouped averages and reset placement

**Use when:** one average per day/subject/category, using all selected records for that category.

The outer loop determines **one output line**; the inner loop gathers the values contributing to that line. Initialize the total inside the outer loop, **before** the inner loop. Calculate/print after the inner loop, still inside the outer loop. Divide by the number actually included. This template assumes every selected record has every requested day and at least one record is selected. A threshold filter instead needs its own included counter; it must not blindly divide by the full dictionary size.

[Back to contents](#contents)


```python
selected_records = {
    "North": {"Monday": 80, "Tuesday": 90},
    "South": {"Monday": 100, "Tuesday": 110},
    "East": {"Monday": 90, "Tuesday": 100},
}
days = ["Monday", "Tuesday"]

for day in days:
    total = 0                         # fresh total for THIS day
    for place in selected_records:
        total += selected_records[place][day]
    average = total / len(selected_records)
    print(f"Average on {day}: {average:.2f}")

# Useful invariant: an average is between its contributing minimum and maximum.
for day in days:
    values = []
    for place in selected_records:
        values.append(selected_records[place][day])
    average = sum(values) / len(values)
    assert min(values) <= average <= max(values)
```

## 5. QUICK COPY — Function header, keyword-only defaults, and a while loop

`def rounds_to_goal(start, goal, *, gain=7):`

- `start`, `goal`: required inputs; callers may supply them positionally or by name.
- `gain=7`: the default is 7 if omitted. A default uses **one `=`**.
- A bare `*` marks the boundary: parameters after it are **keyword-only**. Supply `gain=6`, rather than a bare third positional argument.
- Default and keyword-only are separate rules: the default permits omission; the `*` requires a name when supplied.
- Initialize `score` and `rounds` **inside** the function so every call starts fresh. `return` after the loop reports the finished count.
- If the initial score already reaches the goal, `score < goal` is false immediately and zero rounds are returned. A positive gain guarantees progress toward a larger goal.

For multiplicative growth, replace the update with `population *= (1 + growth_rate)`. A 3% rate is `0.03`, not `3`.

[Back to contents](#contents)


```python
def rounds_to_goal(start, goal, *, gain=7):
    if gain <= 0:
        raise ValueError("gain must be positive")
    score = start
    rounds = 0
    while score < goal:
        score += gain
        rounds += 1
    return rounds

print(rounds_to_goal(5, 30))
print(rounds_to_goal(5, 30, gain=6))
print(rounds_to_goal(30, 30))

# Wrong calling form (leave commented): rounds_to_goal(5, 30, 6)
# Correct: rounds_to_goal(5, 30, gain=6)

def years_to_double(growth_rate):
    if growth_rate <= 0:
        raise ValueError("growth_rate must be positive")
    population = 1000
    years = 0
    while population < 2000:
        population *= 1 + growth_rate
        years += 1
    return years

print("Doubling years:", [years_to_double(r) for r in [0.01, 0.02, 0.05, 0.10]])
```

## 6. QUICK COPY — Lambda, column apply, Series name, and Boolean counts

**Lambda structure:** `function_name = lambda parameter: expression`. The expression is returned automatically; there is no `return` keyword. Multiple inputs are comma-separated before the colon.

**Column apply:** select a Series first: `df["column"].apply(function_name)`. Pass the function **without parentheses**; `.apply()` calls it once per value. The lambda's parameter is one element, not the whole Series.

**Series label:** `result.name = "new_label"`. The Python variable name and the Series `.name` attribute are different things. Changing the Series label does not rename the original DataFrame column.

**Boolean counts:** a comparison already returns `True` / `False`. Summing the Boolean Series counts its `True` entries. Check inclusive wording: `at least` means `>=`.

Jupyter automatically displays the last expression in a cell; use `display(...)` or `print(...)` to show several results.

[Back to contents](#contents)


```python
import pandas as pd

catalog = pd.DataFrame({"item": ["lamp", "bag", "speaker"],
                        "original_price": [32.50, 50.00, 80.00]})

discount = lambda price: round(price * 0.8, 2)
discounted_price = catalog["original_price"].apply(discount)
discounted_price.name = "discounted_price"

at_least_40 = lambda price: price >= 40
qualifies = discounted_price.apply(at_least_40)

print(discounted_price)
print(qualifies)
print("Number qualifying:", qualifies.sum())

# Full def is appropriate when the prompt explicitly requires if/else.
def status(numeric_grade):
    if numeric_grade >= 55:
        return "status = pass"
    else:
        return "status = fail"

grades = pd.DataFrame({"numeric_grade": [85, 55, 45]})
print(grades["numeric_grade"].apply(status))
```

## 7. QUICK COPY — Per-entry try/except with a running total

**Use when:** an order/request has missing keys or invalid numeric strings, and processing should continue for later entries.

1. Initialize the total before the item loop.
2. Put `try/except` inside the loop: one failed entry is handled, then the next iteration starts.
3. Convert one quantity (`int(request[item])`), look up one price (`rates[item]`), then **add** the cost.
4. `KeyError`: missing dictionary key. `ValueError`: text such as `"lots"` cannot be converted to `int`.
5. Use `print(...)` in handlers; a bare string is not an error message display.
6. Return after the whole loop. Defining the function does not call it.

Only the successful `try` path changes the total. If an item has both problems, whichever operation fails first determines the message. This example assumes quantities which successfully convert to `int` are acceptable; add a separate range rule only if the prompt requires it. `int("3.5")` fails; `float("3.5")` succeeds when decimals are allowed.

[Back to contents](#contents)


```python
rates = {"helmet": 3.0, "paddle": 5.0, "life_jacket": 4.0, "dry_bag": 2.0}
request = {"helmet": "2", "fins": "1", "paddle": "several",
           "life_jacket": "3", "dry_bag": "1"}

def rental_total(request, rates):
    total = 0
    for item in request:
        try:
            quantity = int(request[item])
            rate = rates[item]
            total += quantity * rate
        except KeyError:
            print(f"No rental rate for {item}")
        except ValueError:
            print(f"Invalid quantity: {request[item]}")
    return total

print("Total:", rental_total(request, rates))
```

## 8. QUICK COPY — Assertions, expected False, and boundary tests

**Syntax:** `assert condition, "message shown if condition is false"`. All Boolean conditions go **before** the comma. If the condition is true, execution continues without an assertion message. If false, `AssertionError` stops execution.

| Purpose | Placement |
|---|---|
| Precondition: input must satisfy the function's assumptions | first lines **inside** the function |
| Postcondition: calculated result must satisfy a property | **inside**, after result exists, before `return` |
| Test: a known call must give a known result | **outside** the function, below its definition |

In `assert beats_benchmark(3, 5) == False`, the expected answer is `False`. If the actual result is also false, `False == False` is true and the assertion passes. A bare `assert beats_benchmark(3, 5)` would instead demand a truthy result.

Tests should use expected values worked out from the problem, not copied from the code's output. Test below / equal / above a boundary. Use a tolerance for inexact floating-point calculations (`abs(actual - expected) < tolerance`). Assertions check code assumptions/invariants; interactive user validation uses retry logic rather than crashing on typing mistakes.

[Back to contents](#contents)


```python
def beats_benchmark(skill, benchmark):
    return skill > benchmark

# Tests are OUTSIDE the function.
assert beats_benchmark(1.51, 6.72) == False, "lower must not beat benchmark"
assert beats_benchmark(7.94, 6.72) == True, "higher must beat benchmark"
assert beats_benchmark(6.72, 6.72) == False, "equal is not strictly greater"

def shipping_charge(weight_kg):
    assert weight_kg >= 0, "weight must be nonnegative"  # precondition
    if weight_kg <= 2:
        charge = 4
    else:
        charge = 7
    assert charge in [4, 7], "charge must follow the allowed prices"  # postcondition
    return charge

assert shipping_charge(1.9) == 4
assert shipping_charge(2.0) == 4  # catches the silent < versus <= bug
assert shipping_charge(2.1) == 7
assert abs((0.1 + 0.2) - 0.3) < 1e-10
print("All boundary and Boolean tests passed")
```

## 9. Lecture 9 — Dictionary construction, lookup, and mutation

A dictionary maps keys to values: `{key: value, ...}`. Lists use positions; dictionaries use keys. String keys are case sensitive. Integer key `0` means the key 0, not necessarily the first entry.
- Read: `mapping[key]`.
- Replace or add: `mapping[key] = value`. An existing key is overwritten; a new key is inserted.
- Increase using its current value: `mapping[key] = mapping[key] + amount`.
- Values can be numbers, strings, lists, or dictionaries. Keep quotes around literal string keys; an unquoted name is a variable lookup.

[Back to contents](#contents)


```python
scores = {"Mina": 6.5, "Luis": 9.0, "Jo": 4.5}
print(scores["Luis"])
scores["Mina"] = scores["Mina"] + 2
scores["Jo"] = 5.0
scores["Nora"] = 8.0
print(scores)

mixed = {0: "first label", "numbers": [2, 4, 6], "details": {"active": True}}
print(mixed[0], mixed["numbers"][1], mixed["details"]["active"])

new_mapping = {}
new_mapping["example"] = 12
print(new_mapping)
```

## 13. Lecture 9 — Running minimum/maximum with a matching label

Keep TWO pieces of state: the best number so far and the label attached to it. Update both inside the same conditional. For arbitrary finite numeric data, positive/negative infinity avoids assuming a suitable sentinel such as 0.
`>` or `<` keeps the first tied record; `>=` or `<=` keeps the last. Empty input leaves the label `None` here. The NumPy check requires nonempty input.

[Back to contents](#contents)


```python
import numpy as np

records = {
    "Cedar": {"score": 7.1},
    "Harbor": {"score": 8.8},
    "Ridge": {"score": 6.2},
}
best_value = -float("inf")
best_name = None
smallest_value = float("inf")
smallest_name = None

for name in records:
    value = records[name]["score"]
    if value > best_value:
        best_value = value
        best_name = name
    if value < smallest_value:
        smallest_value = value
        smallest_name = name

print("Largest:", best_name, best_value)
print("Smallest:", smallest_name, smallest_value)

flat_scores = {"Cedar": 7.1, "Harbor": 8.8, "Ridge": 6.2}
print("NumPy score check:", np.max(list(flat_scores.values())))
```

## 14. Lecture 9 — Nested loops for grouped totals and means

Choose the outside loop to match the requested output: one result per day means loop over days outside. Reset the total INSIDE that loop but BEFORE the inner record loop. Divide by the number actually selected, not a hardcoded number.
The same structure works for subjects across students or products across stores. Distinguish averaging each field across records from averaging all fields inside one record. This template assumes every selected record contains every requested day.
`round(x, 2)` rounds a number; `f"{x:.2f}"` prints exactly two decimal places.

[Back to contents](#contents)


```python
readings = {
    "North": {"Monday": 20, "Tuesday": 24},
    "South": {"Monday": 26, "Tuesday": 22},
    "East": {"Monday": 23, "Tuesday": 25},
}
selected_names = ["North", "East"]
selected = {}
for name in selected_names:
    selected[name] = readings[name]

days = ["Monday", "Tuesday"]
if len(selected) == 0:
    raise ValueError("Select at least one record.")

for day in days:
    total = 0
    for name in selected:
        total = total + selected[name][day]
    average = total / len(selected)
    print(f"Average on {day}: {average:.2f}")
```

## 16. Lecture 10 — Function anatomy, parameters, and calls

`def` defines reusable work; the body is indented; `return` sends a result to the caller. Defining a function does not run its body. A parameter is the local input name in the definition; an argument is a value supplied in a call.
Arguments can be positional or named using the EXACT parameter names. Caller variable names and result variable names need not match parameter/local names. Put positional arguments before keyword arguments.

[Back to contents](#contents)


```python
def transform(x, y):
    result = y + x**2 + 2*x + 1
    return result

print(transform(2, 3))
print(transform(x=2, y=3))
print(transform(y=3, x=2))

first_input = 2
second_input = 3
answer = transform(first_input, second_input)
print(answer)
```

## 17. Lecture 10 — Local scope and reading globals

Parameters and names assigned inside a function are local to that function. Returning a value does not make the local variable name global; it passes an object back to the caller. A function can read a global name, but passing dependencies as arguments makes it easier to reuse and test.
A pre-existing global variable can hide a scope mistake in Jupyter. Restarting and running in order helps catch hidden state.

[Back to contents](#contents)


```python
global_offset = 10

def reads_global(x):
    local_total = x + global_offset
    return local_total

def takes_dependency(x, offset):
    local_total = x + offset
    return local_total

print(reads_global(3))
print(takes_dependency(3, 10))
try:
    print(local_total)
except NameError:
    print("local_total is not defined outside either function.")
```

## 18. Lecture 10 — Return versus print, early return, and None

`return` ends the CURRENT function call immediately. Statements after that return on the same path will not run. `print` displays a value but does not return that value.

## 22. Lecture 10 — Quadratic roots and a consistent return contract

## 30. Lecture 10 — Birthday simulation: one trial, then estimated probability

Separate one random experiment from the loop that repeats it. Count how many trials contain at least one shared birthday, then divide by the TOTAL trial count.
Assumptions: independent uniformly distributed birthdays on 365 days; no leap day or real-world seasonality. A duplicate among any pair is success; it need not be a match to one particular person. Trials are simulations, so the estimated probability is not exact. A fixed seed makes the example reproducible.

[Back to contents](#contents)


```python
import random

def shared_birthday_trial(room_size, rng):
    seen_birthdays = []
    for person in range(room_size):
        birthday = rng.randint(1, 365)
        if birthday in seen_birthdays:
            return True
        seen_birthdays.append(birthday)
    return False

def birthday_probability(room_size, trials=500):
    if room_size < 0 or trials <= 0:
        raise ValueError("Use a nonnegative room size and positive trial count.")
    rng = random.Random(42)
    successful_trials = 0
    for trial in range(trials):
        if shared_birthday_trial(room_size, rng):
            successful_trials = successful_trials + 1
    return successful_trials / trials

# room_size and trials are assumed to be integers.
for size in [10, 23, 35]:
    print(size, birthday_probability(size))
```

## 31. Lecture 10 — Random stopping: total successes

Track successes and total trials separately. On a failure the TOTAL success count stays unchanged. Return the trial count only after the target is reached.
Assume required is a positive integer and 0 < probability <= 1. The guard raises an error instead of silently returning an unfinished experiment as though it succeeded. This is one random run, so different seeds can give different results.

[Back to contents](#contents)


```python
import numpy as np

def trials_for_total_successes(required, probability=0.25, max_trials=10000):
    if required < 1 or not 0 < probability <= 1:
        raise ValueError("Use a positive target and 0 < probability <= 1.")
    rng = np.random.default_rng(12)
    successes = 0
    trials = 0
    while successes < required:
        if trials >= max_trials:
            raise RuntimeError("Trial limit reached before the target.")
        outcome = rng.choice(["Success", "Failure"], p=[probability, 1-probability])
        trials = trials + 1
        if outcome == "Success":
            successes = successes + 1
    return trials

print(trials_for_total_successes(3))
```

## 32. Lecture 10 — Random stopping: consecutive successes

Consecutive means an uninterrupted run. Increase the current streak on success; RESET that streak to zero on failure. Never reset the total trial count on failure.
The return belongs after the while loop. Compare this template with the total-success version: the important extra branch is the streak reset. Assume required is a positive integer; the trial limit prevents an impractically long run.

[Back to contents](#contents)


```python
import numpy as np

def trials_for_consecutive_successes(required, probability=0.8, max_trials=10000):
    if required < 1 or not 0 < probability <= 1:
        raise ValueError("Use a positive target and 0 < probability <= 1.")
    rng = np.random.default_rng(21)
    streak = 0
    trials = 0
    while streak < required:
        if trials >= max_trials:
            raise RuntimeError("Trial limit reached before the target.")
        outcome = rng.choice(["Success", "Failure"], p=[probability, 1-probability])
        trials = trials + 1
        if outcome == "Success":
            streak = streak + 1
        else:
            streak = 0
    return trials

print(trials_for_consecutive_successes(3))
```

## 38. Lecture 13 • PRIORITY: assert, preconditions, postconditions

**Syntax:** `assert condition_that_should_be_true, "message if false"`.

True: continues silently. False: raises AssertionError with the message. Put BOTH conditions before the comma:
`assert N >= 0 and N < len(roster), "Invalid N"`.
Equivalent: `assert 0 <= N < len(roster), "..."`.

**Placement**
- Precondition: first indented lines inside function, before calculations.
- Postcondition: after calculating the result, before return.
- Tests: outside function, aligned with def, calling known inputs.

`N < len(roster)` compares numbers. `chosen_skill in all_skills` checks membership. `len(roster)` is a number, so `N in len(roster)` is wrong. Build the skills list first; do not test chosen_skill before assigning it. Statements after return do not execute.

Use try/except plus if checks for expected invalid external inputs; use assert for code assumptions and tests. Assertions can be disabled by optimized Python execution, so required external-input validation should use explicit checks.

[Back to contents](#contents)


```python
def checked_selection(roster, order, N):
    assert type(N) == int, "N must be an integer"
    assert 0 <= N < len(order), f"N={N} must leave someone to select"
    assert len(order) == len(roster) and set(order) == set(roster), "Invalid order"
    all_skills = []
    for name in roster:
        all_skills.append(roster[name]["lab_skills"])

    benchmark = float("-inf")
    for position in range(N):
        skill = roster[order[position]]["lab_skills"]
        if skill > benchmark:
            benchmark = skill

    chosen_name = order[-1]
    for position in range(N, len(order)):
        name = order[position]
        if roster[name]["lab_skills"] > benchmark:
            chosen_name = name
            break
    chosen_skill = roster[chosen_name]["lab_skills"]

    assert chosen_name in roster, "Chosen name must exist"
    assert chosen_skill in all_skills, "Chosen skill must appear in roster"
    assert chosen_skill == roster[chosen_name]["lab_skills"], "Name and skill must match"
    return chosen_name, chosen_skill

roster = {"Alex": {"lab_skills": 3.0}, "Sam": {"lab_skills": 8.0}, "Jo": {"lab_skills": 6.0}}
assert checked_selection(roster, ["Alex", "Sam", "Jo"], 1) == ("Sam", 8.0)
try:
    checked_selection(roster, ["Alex", "Sam", "Jo"], -5)
except AssertionError as error:
    print("Expected rejection:", error)
```

## 40. Lecture 13 • PRIORITY: pair-loop bug and mathematical tests

For distinct unordered pairs, start j at i+1:
`for i in range(len(values)):` then `for j in range(i + 1, len(values)):`.

For three people, genuine pairs are (0,1), (0,2), (1,2). Starting j at i includes self-pairs, falsely making every nonempty room contain a birthday match. range includes start and excludes stop. Return True immediately on a match; return False AFTER both loops.

Good tests of random code need not depend on luck. One person cannot match another; 361 people must share one of only 360 dates. Two people's true match probability is 1/360, but a finite sample estimate can be zero. Do not assert a random estimate equals the exact probability.

[Back to contents](#contents)


```python
import random

def has_shared_birthday(room):
    for i in range(len(room)):
        for j in range(i + 1, len(room)):
            if room[i] == room[j]:
                return True
    return False

assert has_shared_birthday([(1, 10)]) == False
assert has_shared_birthday([(1, 10), (2, 10)]) == False
assert has_shared_birthday([(1, 10), (1, 10)]) == True

def one_birthday_trial(room_size, rng):
    assert type(room_size) == int and room_size >= 1
    room = []
    for _ in range(room_size):
        room.append((rng.randint(1, 12), rng.randint(1, 30)))
    return has_shared_birthday(room)

rng = random.Random(13)
assert one_birthday_trial(1, rng) == False
assert one_birthday_trial(361, rng) == True
print("Birthday tests passed.")
```

## 42. Lecture 13 • Random tests, tolerances, and the hidden break bug

More trials reduce typical Monte Carlo sampling error, but a particular estimate is NOT guaranteed to get closer whenever you add trials. Finite statistical tests can fail by chance. Independent probability estimates may wiggle instead of increasing perfectly with room size. A fixed random seed makes examples reproducible; it does not prove the algorithm correct.

The lecture's N=0 test with benchmark initialized to 0 selects the first candidate only because EVERY original skill is greater than 0. True zero scores need special thought. A -infinity benchmark represents no observed candidate and supports arbitrary finite real scores, if consistent with the problem.

**Missing break:** the first qualifying candidate becomes the LAST qualifying candidate. Their skill averages can look similar, hiding a strategy violation. Return chosen_position as evidence or accept a fixed order. For all-positive scores and N=0, the correct lecture strategy must stop at position0. Test a known order with multiple qualifiers; the first must win. Three returned values are unpacked as `name, skill, position = function(...)`.

[Back to contents](#contents)


```python
import random

def choose_from_order(scores, N):
    assert 0 <= N < len(scores)
    benchmark = float("-inf")
    for position in range(N):
        benchmark = max(benchmark, scores[position])
    chosen_position = len(scores) - 1
    for position in range(N, len(scores)):
        if scores[position] > benchmark:
            chosen_position = position
            break
    return scores[chosen_position], chosen_position

assert choose_from_order([3, 8, 9], 1) == (8, 1)
assert choose_from_order([3, 8, 9], 0) == (3, 0)
assert choose_from_order([9, 3, 8], 1) == (8, 2)  # fallback

rng = random.Random(13)
population = [2.0, 4.0, 6.0, 8.0]
sample = []
for _ in range(500):
    sample.append(rng.choice(population))
population_mean = sum(population) / len(population)
sample_mean = sum(sample) / len(sample)
difference = abs(sample_mean - population_mean)
print("Population / sample mean:", population_mean, sample_mean)
print("Within illustrative tolerance 0.5:", difference < 0.5)
# Finite unseeded statistical tolerance tests are not absolute guarantees.
```

## 46. BONUS F / Lab 5 — Two point tuples, polar conversion, and helper reuse

This material came from practice quiz F/Lab 5; it supplements the Lecture 9–13 notes.

Cartesian `(x, y)` stores horizontal/vertical coordinates. Polar `(r, theta)` stores radius/angle. **Each point contains both components**: `point1` is not the radius while `point2` is the angle.

For each point separately, `x = r * np.cos(theta)` and `y = r * np.sin(theta)`. NumPy trig functions use **radians**; `np.pi/2` is 90 degrees, and `np.deg2rad(degrees)` converts degrees if a problem explicitly supplies them.

`polar=False` means the tuples already contain `(x, y)`. With `polar=True`, convert both tuples first. The supplied distance helper needs two coordinate pairs, not two scalar numbers. `return euclidean_dist(...)` passes the helper's numeric result back to the caller. A bare call with no `return` would discard the result and leave an implicit `None`.

[Back to contents](#contents)


```python
import numpy as np

def euclidean_dist(point1, point2):
    x1, y1 = point1
    x2, y2 = point2
    return np.sqrt((x1 - x2)**2 + (y1 - y2)**2)

def cord_dist(point1, point2, *, polar=False):
    if polar:
        r1, theta1 = point1
        r2, theta2 = point2
        point1 = (r1 * np.cos(theta1), r1 * np.sin(theta1))
        point2 = (r2 * np.cos(theta2), r2 * np.sin(theta2))
    return euclidean_dist(point1, point2)

print("Cartesian:", cord_dist((0, 0), (3, 4)))
print("Polar:", cord_dist((3, 0), (4, np.pi/2), polar=True))
assert abs(cord_dist((3, 0), (4, np.pi/2), polar=True) - 5) < 1e-10
# A third positional Boolean is rejected; name it: polar=True.
```

## 47. BONUS Lab 5 — Bare *, *args, and **kwargs are different

| Signature notation | Meaning |
|---|---|
| `def f(a, *, option=False):` | bare `*`: arguments after it must be named |
| `def f(*args):` | collect extra positional arguments into a **tuple** |
| `def f(**kwargs):` | collect extra keyword arguments into a **dictionary** |

`args` and `kwargs` are conventional names; the stars provide the behavior. Collected arguments do not automatically validate types or use every provided value. No arguments produces an empty tuple/dictionary. In calls, `*values` unpacks a sequence into positional arguments; `**settings` unpacks a dictionary into keyword arguments. These extra Lab 5 features were not central to our mixed practice, but help distinguish the bare `*` in F.

Positional arguments normally precede explicit keyword arguments: `f(1, b=2)` is valid for an appropriate signature; `f(a=1, 2)` is a syntax error. A parameter cannot receive two values in one call.

[Back to contents](#contents)


```python
def collect_inputs(*args, **kwargs):
    return args, kwargs

print(collect_inputs(10, "hello", name="Alex", major="Biology"))
print(collect_inputs())

def add(a, b):
    return a + b

numbers = (3, 4)
settings = {"a": 3, "b": 4}
print(add(*numbers))
print(add(**settings))
```
